# `FreeFermion.linalg`

A covariance `Gamma` holds the two-point functions of the Majoranas of a Gaussian state, and everything in this notebook is built from it: a random pure state, Williamson's theorem for the normal modes, and the Wick/Pfaffian contraction that turns an ordered product of Majoranas into a number. The last section tests the Gram matrix of the Majorana-excitation basis on mixed states, where the basis is neither orthogonal nor complete.


```python
import torch

from base import CUDA, COMPLEX, REAL
from FreeFermion.linalg import (RandPureCov, is_pure, majorana_gram, purity,entropy,
                                wick, williamson)
from FreeFermion.cuda import fermion
```

| function | what it returns |
|---|---|
| `RandPureCov(dim)` | random pure covariance `Gamma = q J q^T` of `dim` Majoranas, so `Gamma^2 = -1` |
| `williamson(gamma)` | `(R, lambdas)` with `Gamma = R^T D R`, `D = diag(lambda_k J)`; the state is pure exactly when every `lambda_k = 1` |
| `occupations(covariance)` | the mode occupations `p_k = (1 - lambda_k) / 2` |
| `is_pure(covariance)` | the test `Gamma^2 = -1` |
| `purity(covariance)` | `Tr(rho^2)`, a product of the binary purities of the modes |
| `wick(covariance, operators)` | Gaussian expectation of an ordered Majorana product, by Wick's theorem |
| `majorana_gram(covariance)` | Gram matrix of the Majorana-excitation basis over the Majoranas of the covariance |
| `FreeFermion.cuda.fermion` | the batched Wick contraction chain (`operator_vectors`, `vacuum_contraction`, `pfaffian`) of `FreeFermion/cuda/fermion.cu` |


## 1. Sampling and the Williamson decomposition

The reconstruction error is at round-off, and the eigenvalues are all 1, so the random draw is a pure state: the normal modes are empty.


```python
A = RandPureCov(6)                          # pure covariance of 6 Majoranas, i.e. 3 modes
R, lambdas = williamson(A)

J = torch.tensor(
    [[0.0, -1.0],
     [1.0, 0.0]],
    dtype = A.dtype,
    device = CUDA)
D = torch.block_diag(*[lambdas[k] * J for k in range(3)])   # the D = diag(lambda_k J) of the convention
eye = torch.eye(6, dtype = R.dtype, device = CUDA)

print(f'|Gamma - R^T D R| = {float((A - R.T @ D @ R).abs().max()):.3e}')
print(f'|R^T R - I|       = {float((R.T @ R - eye).abs().max()):.3e}')
print(f'lambdas           = {[round(float(x), 12) for x in lambdas]}')
print(f'is_pure(Gamma)    = {is_pure(A)}')
```

## 2. Normal modes

`coeff` collects the `n` mode coefficients `d_k` over the `2n` Majoranas, and `2 d_k d_k^dag = 1` is the emptiness of mode `k`.


```python
occupations = (1.0 - lambdas) / 2.0                    # p_k of the convention
coeff = (R[0::2] + 1j * R[1::2]) / 2.0                 # d_k over the 2n Majoranas, shape (3, 6)

print(f'occupations p_k          = {[round(float(x), 12) for x in occupations]}')
print(f'purity                   = {purity(A):.15f}')
print(f'|2 c c^dag - I|           = '
      f'{float((2.0 * (coeff @ coeff.conj().T) - torch.eye(3, dtype = coeff.dtype, device = CUDA)).abs().max()):.3e}')
```

## 3. The Wick/Pfaffian chain (CUDA kernels)

The expectation of an ordered product of Majoranas is the Pfaffian of the contraction matrix, and the kernel computes it directly. The 4x4 case is small enough to compare against the recursion that enumerates the pairings by hand; section 5 wraps the same kernel in `wick`, which takes the covariance directly instead of the mode coefficients.


```python
def by_pairings(matrix: torch.Tensor) -> torch.Tensor:
    '''
    brute-force Pfaffian by enumerating the pairings, only for the 4x4 cross-check.
    Args:
        matrix: complex antisymmetric matrix, shape (m, m) with m even.
    Returns:
        The Pfaffian of matrix, a 0-dim tensor.
    '''
    size = matrix.shape[0]
    if size == 0:
        return torch.ones((), dtype = matrix.dtype, device = matrix.device)
    total = torch.zeros((), dtype = matrix.dtype, device = matrix.device)
    for j in range(1, size):
        keep = [k for k in range(1, size) if k != j]
        total = total + (-1) ** (j + 1) * matrix[0, j] * by_pairings(matrix[keep][:, keep])
    return total


antisymmetric = torch.randn(4, 4, dtype = COMPLEX, device = CUDA)
antisymmetric = antisymmetric - antisymmetric.T
print('kernel pfaffian =', complex(fermion.pfaffian(antisymmetric.unsqueeze(0))[0]))
print('by pairings     =', complex(by_pairings(antisymmetric)))
```

```python
# one empty mode: <0|gamma_0 gamma_1|0> = 1j
single = torch.tensor(
    [[0.0, -1.0],
     [1.0, 0.0]],
    dtype = torch.float64,
    device = CUDA)
R, lambdas = williamson(single)
coeff = (R[0::2] + 1j * R[1::2]) / 2.0

# the kernels are batched, so a single operator list is a batch of one (kind 2 = gamma_mu,
# 2 majoranas of 1 mode, hence the 2-vectors and the 2x2 contraction matrix)
codes = torch.tensor(
    [[[2, 0], [2, 1]]],
    dtype = torch.long,
    device = CUDA)
vectors = fermion.operator_vectors(coeff, codes)      # (1, 2, 2)
matrices = fermion.vacuum_contraction(vectors, 1)     # (1, 2, 2)
values = fermion.pfaffian(matrices)                   # (1,)

print('gamma_0 in the mode basis =', vectors[0, 0].tolist())
print('<0|gamma_0 gamma_1|0>     =', complex(values[0]), '(expected 1j)')
```

## 4. Batched calls, which is what the kernels are for

The same operator list repeated over a batch comes back as a batch of values from one call, and the batched form is two orders of magnitude cheaper per value than the call-per-list form.


```python
# 8 identical operator lists [gamma_0, gamma_1], one call for the whole batch
codes = torch.tensor(
    [[[2, 0], [2, 1]]] * 8,
    dtype = torch.long,
    device = CUDA)
vectors = fermion.operator_vectors(coeff, codes)      # (8, 2, 2)
matrices = fermion.vacuum_contraction(vectors, 1)     # (8, 2, 2)
values = fermion.pfaffian(matrices)                   # (8,)

print('shapes:', tuple(vectors.shape), tuple(matrices.shape), tuple(values.shape))
print('values:', values.tolist())
```

```python
import time


def bench(fn, reps: int) -> float:
    '''
    mean wall time of one call of fn in microseconds.
    Args:
        fn: the callable to time, already bound to its arguments.
        reps: number of timed calls after two warm-up calls.
    Returns:
        The mean time of one call in microseconds.
    '''
    for _ in range(2):
        fn()
    torch.cuda.synchronize()
    start = time.perf_counter()
    for _ in range(reps):
        fn()
    torch.cuda.synchronize()
    return (time.perf_counter() - start) / reps * 1e6


many = torch.tensor(
    [[[2, 0], [2, 1]]] * 256,
    dtype = torch.long,
    device = CUDA)
one = many[:1]
batched_us = bench(lambda: fermion.pfaffian(fermion.vacuum_contraction(
    fermion.operator_vectors(coeff, many), 1)), 20)
scalar_us = bench(lambda: [fermion.pfaffian(fermion.vacuum_contraction(
    fermion.operator_vectors(coeff, one), 1)) for _ in range(256)], 2)

print(f'{"form":<22}{"total us":>10}{"per list us":>13}')
print(f'{"batched, 256 lists":<22}{batched_us:>10.1f}{batched_us / 256:>13.2f}')
print(f'{"one by one, 256":<22}{scalar_us:>10.1f}{scalar_us / 256:>13.2f}')
print(f'{"speedup":<22}{scalar_us / batched_us:>9.0f}x')
```

## 5. `wick`: any ordered Majorana product

`wick(covariance, operators)` is the primitive the contractions of this module are built from: the Gaussian expectation of an ordered product of Majoranas, which Wick's theorem turns into the Pfaffian of the contraction matrix. The labels are 0-based Majorana indices of the covariance, a plain list is one product, and a rectangular batch gives a batch of values. A repeated Majorana is its own square, so it contracts with itself: the expectation of `gamma_a gamma_a` is 1, and no product has to be reduced first. A product of an odd number of Majoranas vanishes, and so does the empty one but with the value 1.


```python
Gamma = RandPureCov(16)                       # pure covariance of 8 modes = 16 Majoranas

print(f'{"product":<28}{"wick":>24}{"expected":>12}')
for labels, expected in (([0, 0], 1 + 0j), ([0, 1], -1j * Gamma[0, 1]),
                       ([0, 1, 0, 1], -1 + 0j), ([0, 1, 2], 0j),
                       ([], 1 + 0j)):
    product = 'gamma_' + ' gamma_'.join(str(x) for x in labels) if labels else 'empty product'
    print(f'{product:<28}'
          f'{complex(wick(Gamma, labels)):>24.12g}{complex(expected):>12.6g}')
```

```python
lists = [[0, 1], [2, 3], [4, 5], [1, 4]]         # a batch needs equal lengths
batched = wick(Gamma, lists)
one_by_one = torch.stack([wick(Gamma, one) for one in lists])

print('batched   :', [f'{x.real:.12g}{x.imag:+.12g}j' for x in batched])
print('one by one:', [f'{x.real:.12g}{x.imag:+.12g}j' for x in one_by_one])
print('difference:', float((batched - one_by_one).abs().max()))
```

## 6. `majorana_gram`: the Gram of the excitation basis

The basis states are the products `gamma_S` applied to the Gaussian state, for every subset S of the Majoranas of the given covariance, so $2r$ Majoranas give $2^{2r}$ states. The inner product is defined by $\langle\gamma_{T}^\dagger\gamma_S\rangle$, where the average is taken with respect to the Gaussian state $\Gamma$. 

```python
Gamma = RandPureCov(16) 
print(f'{"Majoranas":>10}{"states":>9}{"rank":>7}{"min eigenvalue":>16}   block structure')
for count in (2, 4, 6, 8, 10, 12):
    G = majorana_gram(Gamma[:count, :count])
    eigenvalues = torch.linalg.eigvalsh(G)
    rank = int((eigenvalues > 1e-9).sum())
    print(f'{count:>10}{G.shape[0]:>9}{rank:>7}{float(eigenvalues.min()):>16.2e}'
          f'   {rank} of {G.shape[0]} states independent')
```

It is worth noting that for a pure state $\Gamma$, the actual rank of the Gram matrix is $2^r$. This has a simple interpretation. Via canonical transformation one can regroup the Majoranas $\gamma$ into Dirac fermions $c_i^\dagger$ and $c_i$. There will be $r$ of these Dirac pairs. Hence, for a pure state $|\Gamma\rangle$, the actual space is spanned by the product of creation operators: $c_i^{\dagger} \dots c_{i'}^\dagger |\Gamma\rangle$, where as the annihilation operators does not appear. 

## 7. Building the Gram on mixed states

The rank $2^r$ derived above holds for a pure covariance. The mixed states here are sampled by slicing a pure covariance: keeping the leading $n$ Majoranas of a pure covariance of $N$ Majoranas gives the valid block covariance $\Gamma_{\text{block}} = \Gamma_{\text{pure}}[\,0{:}n,\,0{:}n\,]$, which still obeys $\Gamma^2 = -1$ only when the block is unentangled with the rest.

The basis is still the $2^n$ products $\gamma_S$, and the Gram is the matrix of $\operatorname{Tr}(\rho\,\gamma_S^\dagger \gamma_T)$. Because $\rho = \bigotimes_k \rho_k$ factorizes over the $r = n/2$ normal modes, this matrix is the Kronecker product of the $r$ local Grams of the operators $\{1, \gamma_{2k}, \gamma_{2k+1}, \gamma_{2k}\gamma_{2k+1}\}$:

* a pure mode, $p_k \in \{0, 1\}$, has $\rho_k$ of rank one, so $\operatorname{Tr}(\rho_k A^\dagger B) = \langle A\psi_k \mid B\psi_k\rangle$ factors through the two-dimensional image, local rank $2$;
* a mixed mode, $0 < p_k < 1$, has $\rho_k$ invertible, so $A \mapsto A\rho_k^{1/2}$ is injective on the four-dimensional operator space, local rank $4$.

The Kronecker product multiplies the ranks, so

$$\operatorname{rank} G = 2^{\,r + m}, \qquad m = \#\{\,k : 0 < p_k < 1\,\},$$

which is the pure $2^r$ at $m = 0$ and the full $2^{2r} = 2^n$ at $m = r$. The table slices a fixed block of $n = 8$ Majoranas out of pure covariances of growing size $N$: the larger the rest, the more of the block's modes are mixed, and the rank climbs from $2^r$ to $2^n$.


```python
block = 8                                             # Majoranas of the basis, so a 256 x 256 Gram
print(f'{"N":>4}{"purity":>18}{"entropy":>18}{"mixed modes":>13}{"rank":>8}{"2^(r+m)":>9}')
for total in (block, block + 2, block + 4, block + 8, 4 * block):
    covariance = RandPureCov(total)[:block, :block]    # mixed block sampled by slicing the pure state
    _, lambdas = williamson(covariance)
    modes = (1.0 - lambdas) / 2.0                      # occupations p_k of the block, so 0 < p_k < 1 means mixed
    mixed = int(((modes > 1e-9) & (modes < 1.0 - 1e-9)).sum())
    rank = int((torch.linalg.eigvalsh(majorana_gram(covariance)) > 1e-9).sum())
    print(f'{total:>4}{purity(covariance):>18.15f}{entropy(covariance):>18.15f}'
          f'{mixed:>13}{rank:>8}{2 ** (block // 2 + mixed):>9}')
```

For large $n$ the Gram has $2^n$ rows and cannot be built, but $r + m$ only needs the $n/2$ normal modes of the block, which an $n \times n$ diagonalisation gives. The grid below estimates it for larger block sizes, including even numbers that are not powers of two (evenness of the Majorana count is the only constraint), one random pure state per row sliced at growing $n$.

The estimate is exactly $\min(n,\,N/2)$: the block spans $2^n$ operators while the whole state has only $2^{N/2}$ dimensions, and the map $A \mapsto (A \otimes I)|\Psi\rangle$ saturates the smaller bound. So $r + m$ grows with $n$ and saturates at $n$ once $N \geq 2n$.

For one large state the same slices show the purity and entropy falling as the block grows, and for the block sizes still affordable to build, the measured Gram rank is exactly $2^{r+m}$ — again at non-power-of-two $n$.


```python
# r + m is read off the n/2 normal modes of the block, so sizes far beyond the 2^n-row Gram are affordable
ns = (6, 10, 12, 20, 24, 40, 48)                       # block sizes in Majoranas
Ns = (12, 20, 24, 40, 48, 56, 96)
print(f'{"N":>4}' + ''.join(f'{n:>5}' for n in ns))
matches = True
for total in Ns:
    pure = RandPureCov(total)                          # one pure state per row, sliced at every block size
    cells = []
    for block in ns:
        if block > total:
            cells.append('    .')
            continue
        modes = (1.0 - williamson(pure[:block, :block])[1]) / 2.0
        mixed = int(((modes > 1e-9) & (modes < 1.0 - 1e-9)).sum())
        matches &= block // 2 + mixed == min(block, total // 2)
        cells.append(f'{block // 2 + mixed:>5}')
    print(f'{total:>4}' + ''.join(cells))
print(f'r + m == min(n, N/2) in every cell: {matches}')
```

```python
# one large state, the same non-power-of-two block sizes: purity and entropy next to r + m
total = 96
print(f'{"n":>4}{"purity":>20}{"entropy":>20}{"mixed modes":>13}{"r + m":>7}')
pure = RandPureCov(total)
for block in ns:
    covariance = pure[:block, :block]
    modes = (1.0 - williamson(covariance)[1]) / 2.0
    mixed = int(((modes > 1e-9) & (modes < 1.0 - 1e-9)).sum())
    print(f'{block:>4}{purity(covariance):>20.15f}{entropy(covariance):>20.15f}'
          f'{mixed:>13}{block // 2 + mixed:>7}')
```
